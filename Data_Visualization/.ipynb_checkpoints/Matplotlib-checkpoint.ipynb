```python
import matplotlib.pyplot as plt 
import numpy as np 
import pandas as pd 
```

**Matplotlib Basics**

```python
x= np.linspace(0,5,11)
y= x**3
```

```python
plt.plot(x,y)
plt.title('Hello')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

**Sub-Plots**

```python
plt.subplot(2,2,1)
plt.plot(x,y)
x=np.linspace(1,4,9)
y=x**(1/3)
plt.subplot(2,2,2)
plt.plot(x,y)
x=np.linspace(-2,1,12)
y=x**2
plt.subplot(2,2,3)
plt.plot(x,y)
x=np.linspace(1,2,11)
y=x**0.5
plt.subplot(2,2,4)
plt.plot(x,y)
```

**Object Oriented way**

```python
fig = plt.figure()
axis1=fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2=fig.add_axes([0.2,0.2,.2,.2])
x1=np.linspace(-3,1,9)
y1=x1**3
axis2.plot(x1,y1)
```

**Types of Plots**

```python
plt.scatter(x,y)
```

```python
y1=[[.5,2,2.5,3.4,4.7,6],[0.3,2.3,4.3,5.4,5.9,6.8]]
plt.hist(y1)
```

```python
plt.boxplot(y1)
```

**Saving Plots**

```python
fig = plt.figure()
ax=fig.add_axes([0.1,0.2,.6,.7])
ax.plot(x,y)
plt.savefig('basicplot.png')
```
