```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import warnings 
warnings.filterwarnings('ignore')
```

```python
df=pd.read_csv('train.txt',sep=';',header= None,names=['text','emotion'])
```

```python
unique_emotions=df['emotion'].unique()
emotion_numbers={}
ind=0
for emo in unique_emotions:
    emotion_numbers[emo]=ind 
    ind+=1
df['emotion']=df['emotion'].map(emotion_numbers)
```

```python
df.head()
```

```python
df['text']=df['text'].apply(lambda x:x.lower())
```

```python
def punc_remover(text):
    s=''
    for i in text: 
        if i not in '''.?!,:;-—()[]{}'"…"''':
            s+=i 
    return s
df['text']=df['text'].apply(punc_remover)
```

```python
def num_remover(text): 
    s=''
    for i in text:
        if not i.isdigit(): 
            s+=i 
    return s 
df['text']=df['text'].apply(num_remover)
```

```python
def emoji_remover(text):
    s=''
    for i in text: 
        
        if i==' ' or (ord(i)<=ord('z') and ord(i)>=ord('A')):
            s+=i 
    return s 
df['text']=df['text'].apply(emoji_remover)
```

```python
stop_words=set(stopwords.words('english'))
stop_words
```

```python
def remove_stpwrds(text): 
    l=[]
    words=text.split()
    for i in words: 
        if not i in stop_words:
            l.append(i)
    return ' '.join(l)
df['text']=df['text'].apply(remove_stpwrds)
```
