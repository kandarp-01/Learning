```python
def plotting(var,num):
    plt.figure(figsize=(8,8))
    plt.subplot(4,2,num)
    sns.histplot(df[var],kde=True)
numeric_columns=['Age','RestingBP','Cholesterol','FastingBS','MaxHR','Oldpeak','HeartDisease']
for key,val in enumerate(numeric_columns): 
    plotting(val,key+1)
```
