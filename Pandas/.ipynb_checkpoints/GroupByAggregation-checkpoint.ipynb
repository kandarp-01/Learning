```python
import numpy as np 
import pandas as pd 
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
```

```python
df
```

```python
a=df.groupby('Category')['Sales'].sum()
```

```python
a
```

```python
a=df.groupby('Store')['Sales'].sum()
```

```python
a
```

```python
a=df.groupby('Store')[['Sales','Quantity']].sum()
```

```python
a['Sales']*a['Quantity']
```

```python
a=df.groupby(['Store','Category'])[['Sales','Quantity']].sum()
```

```python
a['Sales']*a['Quantity']
```


**Aggregation**

```python
# min, max, mean, median, mode, count, std 
```

```python
df['Sales'].mean()
```

```python
df['Sales'].median()
```

```python
df['Sales'].min()
```

```python
df['Sales'].max()
```

```python
df[df['Sales']>200]['Sales'].count()
```

```python
df['Sales'].std()
```

```python
df['Sales'].mode()
```

```python
df['Sales'].agg(['mean','median','min','max','count'])
```
