```python
import numpy as np 
import pandas as pd  
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df=sns.load_dataset('tips')
```

```python
df
```

```python
sns.countplot(x=df['sex'],hue=df['smoker'])
```

```python
sns.barplot(x=df['sex'],y=df['total_bill'])
```

```python
sns.barplot(x=df['sex'],y=df['total_bill'],estimator=np.max)
```

```python
sns.boxplot(x='tip',y='day',data=df,hue='sex',palette='rainbow')
```

```python
sns.violinplot(x='day',y='total_bill',data=df,hue='sex',palette='rainbow')
```

```python
sns.stripplot(x='day',y='tip',data=df)
```

```python
sns.stripplot(x='day',y='tip',data=df,hue='sex')
```

```python
sns.swarmplot(x='day',y='tip',data=df)
```

```python
sns.swarmplot(x='day',y='tip',data=df,hue='sex')
```
