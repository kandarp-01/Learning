```python
import numpy as np 
from scipy import stats
```

```python
grp_A=[85,88,90,92,87,85,89,91,86,88]
grp_B=[82,84,80,83,81,79,78,85,84,83]
t,p=stats.ttest_ind(grp_A,grp_B,equal_var=False)
meanA=np.mean(grp_A)
meanB=np.mean(grp_B)
stdA=np.std(grp_A)
stdB=np.std(grp_B)
n1=len(grp_A)
n2=len(grp_B)
alpha=0.05
```

```python
t_score=(meanA-meanB)/np.sqrt((stdA**2/n1)+(stdB**2/n2))
t_score
t
```

```python
p_value=2*(1-stats.t.cdf(abs(t_score),df=min(n1-1,n2-1)))
p_value
```

```python
if p_value<alpha:
    print("Null hypothesis is rejected")
else:
    print("Null hypothesis is accepted")
```
