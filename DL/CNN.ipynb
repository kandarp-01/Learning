```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
import warnings 
warnings.filterwarnings('ignore')

from sklearn.linear_model import Perceptron
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split 

from sklearn.metrics import accuracy_score,classification_report,confusion_matrix

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Conv2D, Flatten

from tensorflow.keras.layers import MaxPooling2D, Dropout 
from tensorflow.keras.utils import to_categorical
```

```python
df=pd.read_csv('train 2.csv')
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
cols=df.columns
```

```python
c=0
for i in range(1,29):
    for j in range(1,29):
        c+=1
        df.rename(columns={cols[c]:f"{i}x{j}"},inplace=True)
```

```python
df.head()
```

```python
X=df.drop('label',axis=1)
y=df['label']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X.values, y.values, test_size=0.20, random_state=42)
```

```python
#Preprocessing 
X_train=X_train.astype('float32')/255.0
X_test=X_test.astype('float32')/255.0
```

```python
X_train_img=X_train.reshape(-1,28,28)
X_test_img=X_test.reshape(-1,28,28)
```

```python
X_train_img
```

```python
y_train_cat=to_categorical(y_train,10)
y_test_cat=to_categorical(y_test,10)
```

```python
perceptron=Sequential([
    Flatten(input_shape=(28,28)),
    Dense(10,activation='softmax')
])
```

```python
perceptron.compile(optimizer='sgd',loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
hist_perc=perceptron.fit(X_train_img,y_train_cat,epochs=10,validation_data=(X_test_img,y_test_cat),
                         batch_size=32,verbose=1)
```

```python
acc_perc=perceptron.evaluate(X_test_img,y_test_cat,verbose=0)[1]
```

```python
acc_perc
```

```python
ANN=Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128,activation='relu'),
    Dense(64,activation='relu'),
    Dense(10,activation='softmax')
])
```

```python
ANN.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
ANN_hist=ANN.fit(X_train_img,y_train_cat,validation_data=(X_test_img,y_test_cat),epochs=10,verbose=1,batch_size=32)
```

```python
acc_ann=ANN.evaluate(X_test_img,y_test_cat,verbose=0)[1]
```

```python
acc_ann
```

```python
X_train_cnn=X_train.reshape(-1,28,28,1) #last param is for color channel b&w-1, rgb-3
X_test_cnn=X_test.reshape(-1,28,28,1)
```

```python
cnn=Sequential([
    Conv2D(32, kernel_size=(3,3),activation='relu',input_shape=(28,28,1)),
    MaxPooling2D(pool_size=(2,2)),
    Conv2D(64,kernel_size=(3,3),activation='relu'),
    MaxPooling2D(pool_size=(2,2)),
    Flatten(),
    Dense(128,activation='relu'),
    Dropout(0.5),
    Dense(10,activation='softmax')
])
```

```python
cnn.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
hist_cnn=cnn.fit(X_train_cnn,y_train_cat,validation_data=(X_test_cnn,y_test_cat),epochs=5,verbose=1,batch_size=32)
```
